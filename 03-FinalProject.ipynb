## Research Question

How well geographic and demographic factors predict life expectancy between first, second and third world countries across decades from 1960 to today. Do the most important factors change over time? Do some factors lose predictive weight as nations transition between economic tiers?

## Background and Prior Work

**Updated**\
Since the 1950's human life expectancy, has skyrocketed. Most people today have higher life expectancies than the richest countries in 1950 <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). Global health inequality has also diminished significantly since the 1950s; a newborn in some countries in the 1950s could expect to live twice as much as a newborn in other countries. An interesting factor to note is that a big factor in contributing to these higher life expectancies is the higher infant mortality rates in the 1950s that skews the number a bit, however, life expectancy has still increased for all ages regardless <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2).

There exists a lot of prior research on topics of what has helped increase global life expectancy. A study in the Jounrnal of Economics identifies factors that contributed to this as improved nutrition, better public health through government, urbanization, vaccination, and medical improvements. Furthermore, since WW2 (1945), poor countries have gained rapid improvements in life expectancy, with some gaining over one year per year <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). It's interesting to note that this may imply life expectancy correlates with diminishing returns, where a lot of gains can be made from basic quaility of life changes, however it starts plateuing at the top. 

A useful graph to track changes across time is called the Preston Curve. This graph plots life expectancy against GDP. Over time, this curve has been observed to have risen, which matches with what we expect given the higher life expectancy. However, it is equally important to note that while this trend holds as a large, individual countries have been observed displaying the opposite trend at times as well. <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). This brings up an important distinction that while income is a good predictor of health, there are still other factors at play in determining life expectancy.

Furthermore, while the Preston Curve illustrates that GDP is a foundational predictor, its flattening at higher income levels suggests a transition in the underlying drivers of longevity. This project aims to move beyond aggregate economic metrics to identify which specific geographic and demographic factors (such as urbanization rates, education levels, or climate stability) gain predictive weight as nations move up the economic tier. By analyzing these shifts across decades, we seek to determine if the "formula" for increasing life expectancy has fundamentally changed, and whether certain variables that were critical in a 1960s "Third World" context have lost their predictive power in the modern era.

When countries transition into a different economic tier, it can be broadly attributed to a couple of factors using a long time scale. These factors are energy use, trade, capital, labour, human capital development and foreign direct investment <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note5).However interestingly enough in a short time frame, human capitial has a negative impact, while energy usage and foreign direct investment have a positive impact. <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note5). 

And whenever countries shift economic tiers and grow towards the top of the Preston Curve, it can be seen that the factors that affected life expectancy the most stop having such a huge effect while other factors grow in importance. This can be seen with the shift in effect of certain diseases. In developing countries, infectious diseases is what affects life expectancy the most through other factors like access to clean water. But once countries develop more, they increase access to these factors that decreases risk of infectious disease so more people develop chronic illness which then affect thier life expectancy <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note6). As countries shift on the Preston Curve, different factors lose their weight on life expectancy and other factors replace them.

#### Section 1 — Life expectancy trends by development tier (1961–2024)

**What:** Median life expectancy over time, stratified by our GDP-based development tiers.

**Why:** Our research question asks whether longevity gains differ across economic tiers and decades. Medians are used because life expectancy and GDP are right-skewed with occasional crisis-year drops.

```python
import sys
sys.path.append("./modules")

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from eda_utils import build_analysis_panel, tier_order

sns.set_theme(style="whitegrid", context="notebook")

# Build once and save so this section reloads quickly
panel = build_analysis_panel()
panel.to_csv("data/02-processed/analysis_panel.csv", index=False)

panel = pd.read_csv("data/02-processed/analysis_panel.csv")
panel["dev_tier"] = pd.Categorical(panel["dev_tier"], categories=tier_order(), ordered=True)

print(f"{len(panel):,} country-years | 1961-2024")
print(panel["dev_tier"].value_counts(dropna=False))
```

```python
tier_ts = (
    panel.dropna(subset=["dev_tier", "life_expectancy_total"])
    .groupby(["year", "dev_tier"], observed=True)["life_expectancy_total"]
    .median()
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10, 5))
sns.lineplot(
    data=tier_ts,
    x="year",
    y="life_expectancy_total",
    hue="dev_tier",
    hue_order=tier_order(),
    palette="viridis",
    linewidth=2,
    ax=ax,
)
ax.set(
    title="Median life expectancy by development tier",
    xlabel="Year",
    ylabel="Life expectancy at birth (years)",
)
ax.legend(title="Development tier")
plt.tight_layout()
plt.show()
```

```python
snapshots = panel[panel["year"].isin([1970, 2000, 2024])].dropna(subset=["dev_tier", "life_expectancy_total"])

fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(
    data=snapshots,
    x="year",
    y="life_expectancy_total",
    hue="dev_tier",
    hue_order=tier_order(),
    ax=ax,
)
ax.set(
    title="Distribution of life expectancy at three time points",
    xlabel="Year",
    ylabel="Life expectancy at birth (years)",
)
plt.tight_layout()
plt.show()
```

**Interpretation:** All three development tiers gained years of life expectancy since the 1960s, reflecting substantial improvements in global health outcomes over time. Developing countries started at much lower life expectancy levels but experienced the largest absolute gains, while developed countries consistently maintained the highest median life expectancy throughout the period. The gap between developing and developed countries narrowed considerably through the 1990s and 2000s, consistent with the global catch-up effect described by the Preston curve literature. However, the persistent separation between the tier-specific trends suggests that economic development remains strongly associated with longevity. The fact that the lines rarely cross supports using GDP-based development tier as an important stratification variable in later correlation and regression analyses.

**What:** Preston-style plots (life expectancy vs. log GDP), absolute latitude vs. life expectancy, and decade-level correlations among key variables.

**Why:** We need to see whether GDP and geography relate to longevity differently by tier and whether those relationships change over time—central to our question about shifting predictive weight.

```python
recent = panel[panel["year"] >= 2015].dropna(
    subset=["life_expectancy_total", "log_gdp_per_capita", "dev_tier"]
)

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(
    data=recent,
    x="log_gdp_per_capita",
    y="life_expectancy_total",
    hue="dev_tier",
    hue_order=tier_order(),
    alpha=0.5,
    s=35,
    ax=ax,
)
ax.set(
    title="Preston curve snapshot (2015–2024)",
    xlabel="Log GDP per capita (USD)",
    ylabel="Life expectancy at birth (years)",
)
plt.tight_layout()
plt.show()
```

**Interpretation:** Life expectancy increases with log GDP per capita but flattens at higher income levels, demonstrating the classic Preston curve relationship. The steepest gains occur among lower-income countries, where increases in GDP are associated with large improvements in longevity, while gains become smaller among wealthier countries. Developed countries cluster in the upper-right portion of the plot with consistently high life expectancy, whereas developing countries show greater vertical spread at similar GDP levels. This variation indicates that factors beyond GDP may also influence life expectancy such as healthcare systems, conflict, education, or climate may also influence life expectancy and should be considered in later regression analyses.

```python
geo = panel.dropna(subset=["abs_latitude", "life_expectancy_total", "dev_tier"])

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(
    data=geo[geo["year"] >= 2015],
    x="abs_latitude",
    y="life_expectancy_total",
    hue="dev_tier",
    hue_order=tier_order(),
    alpha=0.5,
    s=35,
    ax=ax,
)
ax.set(
    title="Geography proxy: |latitude| vs. life expectancy (2015–2024)",
    xlabel="Absolute latitude (degrees)",
    ylabel="Life expectancy at birth (years)",
)
plt.tight_layout()
plt.show()
```

**Interpretation:** We do not yet have temperature data in the panel, so absolute latitude serves as a coarse geographic proxy for climatic differences between tropical and temperate regions. The relationship between latitude and life expectancy appears weak, with countries exhibiting substantial variation in life expectancy across nearly all latitude ranges. Countries at similar latitudes often exhibit very different life expectancy levels, with developed countries generally clustering at higher values than developing countries across much of the latitude range. This suggests that geography alone cannot explain differences in longevity and motivates adding direct climate variables in future analyses while retaining development-tier stratification in later models.

```python
corr_vars = [
    "life_expectancy_total",
    "log_gdp_per_capita",
    "health_spend_pct_gdp",
    "health_spend_per_capita_usd",
    "abs_latitude",
]
decades = [1960, 1980, 2000, 2020]

fig, axes = plt.subplots(2, 2, figsize=(11, 9))
for ax, decade in zip(axes.flat, decades):
    subset = panel[panel["decade"] == decade][corr_vars].dropna(how="all")
    c = subset.corr(numeric_only=True)
    sns.heatmap(
        c,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        ax=ax,
        cbar_kws={"shrink": 0.8},
    )
    ax.set_title(f"{decade}s")
plt.suptitle("Correlations by decade", y=1.02)
plt.tight_layout()
plt.show()
```

Beginning in the 2000s, health-spending variables become available and show positive correlations with life expectancy, particularly for per-capita spending. Health spending per capita is positively correlated with GDP, indicating that wealthier countries generally spend more on healthcare. These patterns suggest that GDP and healthcare investment will be important predictors in later regression models, while climate variables may help explain additional variation not captured by economic factors alone.
